```python
x=df.iloc[:,1:]
y=df.iloc[:,0] # Corrected: use 'label' column
ohe=OneHotEncoder(handle_unknown='ignore') # Added handle_unknown to avoid errors with unseen categories
y=ohe.fit_transform(np.array(y).reshape(-1,1)).toarray()
y
```

```python
# Get number of output classes from y
num_classes = y.shape[1]

#layer1
w1=np.random.randn(784,100)*.01
b1=np.zeros((1,100))
#layer2
w2=np.random.randn(100, num_classes)*.01 # Corrected: use num_classes
b2=np.zeros((1, num_classes)) # Corrected: use num_classes
```

```python
def relu(x):
  return np.maximum(0,x)
```

```python
def cross_entropy_loss(y_true, y_pred):
    epsilon = 1e-8

    y_pred = np.clip(y_pred, epsilon, 1 - epsilon)

    loss = -np.mean(
        np.sum(y_true * np.log(y_pred), axis=1)
    )
    return loss
```

```python
def relu_der(x):
  return (x>0).astype(float)
```

```python
def softmax(x):
  # Subtract the maximum value for numerical stability, applied per sample
  x_max = np.max(x, axis=1, keepdims=True)
  exp_x = np.exp(x - x_max)
  # Sum along axis=1 to normalize probabilities for each sample
  return exp_x / np.sum(exp_x, axis=1, keepdims=True)
```

```python
def forward(x,w1,b1,w2,b2):
  z1=np.dot(x,w1)+b1
  a1=relu(z1)
  z2=np.dot(a1,w2)+b2
  ypred=softmax(z2) # Corrected: return softmax output directly, not its max value
  return z1,a1,z2,ypred
```

```python
def backward(X, y_true, Z1, A1, A2, W2):

    m = X.shape[0]
    #ouput layer
    dZ2 = (A2 - y_true) / m

    dW2 = np.dot(A1.T, dZ2)

    db2 = np.sum(dZ2, axis=0, keepdims=True)

    #hidden layer

    dA1 = np.dot(dZ2, W2.T)

    dZ1 = dA1 * relu_der(Z1)

    dW1 = np.dot(X.T, dZ1)

    db1 = np.sum(dZ1, axis=0, keepdims=True)


    return dW1, db1, dW2, db2
```

```python
lr=.07 # Increased learning rate
epoch=500 # Increased number of epochs
for i in range(epoch):
  z1,a1,z2,ypred=forward(x,w1,b1,w2,b2)
  loss=cross_entropy_loss(y,ypred)
  dw1,db1,dw2,db2=backward(x,y,z1,a1,ypred,w2)
  w1=w1-lr*dw1
  b1=b1-lr*db1
  w2=w2-lr*dw2
  b2=b2-lr*db2
  print(f"epoch :{i} loss={loss}")
  if (i%10==0):
    pred=np.argmax(ypred,axis=1)
    actual=np.argmax(y,axis=1)
    accuracy=np.mean(pred==actual)
    print(f"accuracy={accuracy}")
```
